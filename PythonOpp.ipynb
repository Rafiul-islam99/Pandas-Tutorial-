```python


# #**Objected Oriented Programming**

# 1. Class and Object
class Animal:
  # 2. Constructor
  def __init__(self, name, species):
    self.name = name  # 3. Attributes
    self.species = species

  # 4. Methods
  def speak(self):
    print(f"{self.name} the {self.species} makes a sound.")

  # 5. Encapsulation: Data and methods are bundled within the class
  # Private attributes (convention, not strict enforcement in Python)
  def _private_method(self):
      print("This is a private method.")

  def get_name(self):
      return self.name

  # 6. Abstraction: Hiding complex implementation details
  # (More apparent in inheritance and interfaces, but methods themselves are forms of abstraction)
  def move(self):
      # This method can have complex logic internally
      print(f"{self.name} is moving.")


# Create an object (instance of the Animal class)
my_animal = Animal("Leo", "Lion")

# Access attributes
print(f"Name: {my_animal.name}")
print(f"Species: {my_animal.species}")

# Call methods
my_animal.speak()
my_animal.move()

# 7. Inheritance: Creating a new class based on an existing class
class Dog(Animal):
  # Inherits attributes and methods from Animal

  # Override the constructor
  def __init__(self, name, breed):
    super().__init__(name, "Dog") # Call the parent class constructor
    self.breed = breed

  # Override the speak method (Polymorphism)
  def speak(self):
    print(f"{self.name} the {self.species} barks!")

  # Add a new method specific to Dog
  def fetch(self):
    print(f"{self.name} is fetching the ball.")

# Create an object of the Dog class
my_dog = Dog("Buddy", "Golden Retriever")

# Access inherited attributes and methods
print(f"Name: {my_dog.name}")
print(f"Species: {my_dog.species}")
my_dog.speak() # Calls the overridden method in Dog
my_dog.move() # Calls the inherited method from Animal

# Access the new method specific to Dog
my_dog.fetch()

# 8. Polymorphism: Ability of objects of different classes to respond to the same method call in their own way
# (Already demonstrated with the 'speak' method in Animal and Dog)

def make_animal_speak(animal):
  animal.speak()

print("\nPolymorphism example:")
make_animal_speak(my_animal) # Calls Animal.speak()
make_animal_speak(my_dog)    # Calls Dog.speak()

# Another example with the original dog class (although it needs a self parameter in __init__)
# Let's correct the original dog class for demonstration purposes
class SimpleDog:
  def __init__(self): # Corrected: __init__ needs self
    pass
  def bark(self):
    print("woof")

# Create an instance and call the method
simple_dog_instance = SimpleDog()
simple_dog_instance.bark() # Now this works correctly

# Note: Calling a method like `dog.bark(2)` directly on the class without an instance and passing
# a non-instance object (like 2) as the first argument will usually lead to an error or unexpected behavior
# unless the method is designed as a class method (`@classmethod`) or a static method (`@staticmethod`).
# The standard way to call an instance method is on an instance: `instance_name.method_name()`.

```
